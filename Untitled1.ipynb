```python

# ============================================================
# UNIT 1 - TASK 2
# PRODUCT REVIEW SENTIMENT ANALYSIS USING MACHINE LEARNING
# Algorithms:
# 1. Decision Tree
# 2. K-Nearest Neighbours (KNN)
# 3. Support Vector Machine (SVM)
# 4. Naive Bayes
# ============================================================


# CELL 1: INSTALL REQUIRED LIBRARIES

!pip -q install datasets pandas numpy scikit-learn matplotlib seaborn nltk


# CELL 2: IMPORT LIBRARIES

import re
import html
import time
import warnings
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from collections import Counter
from nltk.stem import PorterStemmer
from sklearn.feature_extraction.text import (
    TfidfVectorizer, ENGLISH_STOP_WORDS
)
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import LinearSVC
from sklearn.naive_bayes import MultinomialNB

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

from datasets import load_dataset

warnings.filterwarnings("ignore")

print("Libraries imported successfully!")


# CELL 3: DOWNLOAD AMAZON REVIEWS DATASET

print("Downloading Amazon Reviews Polarity Dataset...")

dataset = load_dataset(
    "fancyzhx/amazon_polarity",
    split="train",
    streaming=True
)

# Shuffle the streaming dataset
dataset = dataset.shuffle(
    seed=42,
    buffer_size=50000
)

# Select 5000 positive and 5000 negative reviews
required_per_class = 5000
counts = {0: 0, 1: 0}
records = []

for item in dataset:

    label = int(item["label"])

    if label not in (0, 1):
        continue

    if counts[label] >= required_per_class:
        continue

    title = item.get("title") or ""
    content = item.get("content") or ""

    review = str(title) + " " + str(content)

    records.append({
        "review": review.strip(),
        "label": label
    })

    counts[label] += 1

    if all(
        counts[c] >= required_per_class
        for c in (0, 1)
    ):
        break

if not all(counts[c] >= required_per_class for c in (0, 1)):
    raise RuntimeError(
        "Could not collect enough reviews. Check dataset access."
    )

df = pd.DataFrame(records)

# Convert labels into readable class names
df["sentiment"] = df["label"].map({
    0: "Negative",
    1: "Positive"
})

print("\nDataset downloaded successfully!")
print("Dataset shape:", df.shape)
print("\nFirst five records:")
print(df.head())

# Save dataset
df.to_csv("amazon_reviews_sample.csv", index=False)


# CELL 4: DATASET INFORMATION

print("\nDataset Information")
print("-------------------")

print("Total Reviews:", len(df))
print("Positive Reviews:", counts[1])
print("Negative Reviews:", counts[0])

print("\nClass Distribution:")
print(df["sentiment"].value_counts())

# Class distribution graph
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="sentiment"
)

plt.title("Amazon Reviews - Class Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.tight_layout()
plt.show()


# CELL 5: CHECK MISSING VALUES AND DUPLICATES

print("\nMissing Values:")
print(df.isnull().sum())

df.dropna(subset=["review"], inplace=True)

df.drop_duplicates(
    subset=["review", "label"],
    inplace=True
)

df.reset_index(drop=True, inplace=True)

print("\nDataset after cleaning:")
print(df.shape)


# CELL 6: TEXT PREPROCESSING

stemmer = PorterStemmer()

# Preserve negation words because they are important
# for sentiment classification.
stop_words = set(ENGLISH_STOP_WORDS) - {
    "not", "no", "nor"
}

def preprocess_text(text):

    # Convert to lowercase
    text = str(text).lower()

    # Decode HTML entities
    text = html.unescape(text)

    # Remove HTML tags
    text = re.sub(r"<[^>]+>", " ", text)

    # Extract words
    words = re.findall(r"[a-z]+", text)

    # Remove stopwords and apply stemming
    words = [
        stemmer.stem(word)
        for word in words
        if word not in stop_words and len(word) > 1
    ]

    return " ".join(words)


print("\nPreprocessing reviews...")

df["clean_text"] = df["review"].apply(preprocess_text)

print("\nOriginal Review:")
print(df["review"].iloc[0])

print("\nPreprocessed Review:")
print(df["clean_text"].iloc[0])

# Remove empty reviews after preprocessing
df = df[df["clean_text"].str.strip() != ""].copy()
df.reset_index(drop=True, inplace=True)

print("\nPreprocessing completed!")


# CELL 7: TRAIN-TEST SPLIT

X_text = df["clean_text"]
y = df["label"]

X_train_text, X_test_text, y_train, y_test = train_test_split(
    X_text,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining Samples:", len(X_train_text))
print("Testing Samples:", len(X_test_text))


# CELL 8: TF-IDF FEATURE EXTRACTION

tfidf = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 1),
    min_df=2,
    sublinear_tf=True,
    stop_words="english"
)

# Fit only on training data to prevent data leakage
X_train = tfidf.fit_transform(X_train_text)

# Transform test data using the same vocabulary
X_test = tfidf.transform(X_test_text)

print("\nTF-IDF Feature Extraction Completed!")
print("Training Matrix Shape:", X_train.shape)
print("Testing Matrix Shape:", X_test.shape)
print("Total Vocabulary Features:", len(tfidf.vocabulary_))

# Display some vocabulary terms
print("\nSample TF-IDF Vocabulary:")
print(list(tfidf.vocabulary_.keys())[:20])


# CELL 9: INITIALIZE FOUR CLASSIFIERS

models = {

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=5,
        metric="cosine",
        algorithm="brute",
        n_jobs=-1
    ),

    "SVM": LinearSVC(
        C=1.0,
        random_state=42
    ),

    "Naive Bayes": MultinomialNB(
        alpha=1.0
    )
}


# CELL 10: TRAIN AND EVALUATE ALL FOUR MODELS

results = []
trained_models = {}
predictions_dict = {}

for name, model in models.items():

    print("\nTraining:", name)

    start_time = time.time()

    # Model training
    model.fit(X_train, y_train)

    # Model prediction
    predictions = model.predict(X_test)

    end_time = time.time()

    elapsed_time = end_time - start_time

    # Evaluation metrics
    accuracy = accuracy_score(y_test, predictions)

    precision = precision_score(
        y_test,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        zero_division=0
    )

    results.append({
        "Algorithm": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "Time (seconds)": elapsed_time
    })

    trained_models[name] = model
    predictions_dict[name] = predictions

    print("Accuracy :", round(accuracy, 4))
    print("Precision:", round(precision, 4))
    print("Recall   :", round(recall, 4))
    print("F1 Score :", round(f1, 4))
    print("Time     :", round(elapsed_time, 2), "seconds")


# CELL 11: COMPARATIVE PERFORMANCE TABLE

results_df = pd.DataFrame(results)

print("\n==========================================")
print("COMPARATIVE PERFORMANCE OF FOUR ALGORITHMS")
print("==========================================")

print(
    results_df.round(4).to_string(index=False)
)

# Save comparison table
results_df.to_csv(
    "model_comparison.csv",
    index=False
)


# CELL 12: PERFORMANCE COMPARISON GRAPH

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score"
]

plot_df = results_df.set_index("Algorithm")[metrics]

ax = plot_df.plot(
    kind="bar",
    figsize=(12, 6),
    width=0.8
)

plt.title(
    "Performance Comparison of Machine Learning Algorithms"
)

plt.xlabel("Classification Algorithms")
plt.ylabel("Performance Score")

plt.ylim(0, 1.05)
plt.xticks(rotation=0)
plt.legend(loc="lower right")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.savefig(
    "performance_comparison.png",
    dpi=300
)
plt.show()


# CELL 13: INDIVIDUAL ACCURACY GRAPH

plt.figure(figsize=(9, 5))

sns.barplot(
    data=results_df,
    x="Algorithm",
    y="Accuracy"
)

plt.title("Accuracy Comparison")
plt.xlabel("Algorithm")
plt.ylabel("Accuracy")

plt.ylim(0, 1.05)
plt.xticks(rotation=15)
plt.tight_layout()

plt.savefig(
    "accuracy_comparison.png",
    dpi=300
)
plt.show()


# CELL 14: CONFUSION MATRICES FOR ALL MODELS

fig, axes = plt.subplots(
    2, 2,
    figsize=(12, 10)
)

axes = axes.flatten()

for i, (name, predictions) in enumerate(
    predictions_dict.items()
):

    cm = confusion_matrix(
        y_test,
        predictions,
        labels=[0, 1]
    )

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["Negative", "Positive"],
        yticklabels=["Negative", "Positive"],
        ax=axes[i]
    )

    axes[i].set_title(name)
    axes[i].set_xlabel("Predicted Label")
    axes[i].set_ylabel("Actual Label")

plt.suptitle(
    "Confusion Matrix Comparison",
    fontsize=16
)

plt.tight_layout()

plt.savefig(
    "confusion_matrices.png",
    dpi=300
)

plt.show()


# CELL 15: CLASSIFICATION REPORT

for name, predictions in predictions_dict.items():

    print("\n================================")
    print("CLASSIFICATION REPORT:", name)
    print("================================")

    print(
        classification_report(
            y_test,
            predictions,
            target_names=["Negative", "Positive"],
            zero_division=0
        )
    )


# CELL 16: WHAT-IF EXPERIMENT
# Change TF-IDF max_features parameter
# 1000, 3000, 5000, 10000

feature_values = [1000, 3000, 5000, 10000]

what_if_results = []

print("\n================================")
print("WHAT-IF EXPERIMENT")
print("================================")

for feature_count in feature_values:

    print(
        "\nTesting max_features =",
        feature_count
    )

    # Create a new vectorizer
    vectorizer = TfidfVectorizer(
        max_features=feature_count,
        ngram_range=(1, 1),
        min_df=2,
        sublinear_tf=True,
        stop_words="english"
    )

    # Use the same train-test split
    X_train_new = vectorizer.fit_transform(
        X_train_text
    )

    X_test_new = vectorizer.transform(
        X_test_text
    )

    for name, model_class in [
        ("Decision Tree", DecisionTreeClassifier),
        ("KNN", KNeighborsClassifier),
        ("SVM", LinearSVC),
        ("Naive Bayes", MultinomialNB)
    ]:

        # Create fresh model with the same parameters
        if name == "Decision Tree":
            model = DecisionTreeClassifier(
                random_state=42
            )

        elif name == "KNN":
            model = KNeighborsClassifier(
                n_neighbors=5,
                metric="cosine",
                algorithm="brute",
                n_jobs=-1
            )

        elif name == "SVM":
            model = LinearSVC(
                C=1.0,
                random_state=42
            )

        else:
            model = MultinomialNB(
                alpha=1.0
            )

        model.fit(
            X_train_new,
            y_train
        )

        pred = model.predict(
            X_test_new
        )

        what_if_results.append({

            "Features": feature_count,

            "Algorithm": name,

            "Accuracy": accuracy_score(
                y_test, pred
            ),

            "Precision": precision_score(
                y_test, pred,
                zero_division=0
            ),

            "Recall": recall_score(
                y_test, pred,
                zero_division=0
            ),

            "F1 Score": f1_score(
                y_test, pred,
                zero_division=0
            )
        })


# Convert what-if results into DataFrame

what_if_df = pd.DataFrame(
    what_if_results
)

print("\nWHAT-IF EXPERIMENT RESULTS")

print(
    what_if_df.round(4).to_string(index=False)
)

what_if_df.to_csv(
    "what_if_results.csv",
    index=False
)


# CELL 17: WHAT-IF ACCURACY GRAPH

plt.figure(figsize=(11, 6))

sns.lineplot(
    data=what_if_df,
    x="Features",
    y="Accuracy",
    hue="Algorithm",
    marker="o"
)

plt.title(
    "Effect of TF-IDF Feature Count on Accuracy"
)

plt.xlabel("Number of TF-IDF Features")
plt.ylabel("Accuracy")

plt.ylim(0, 1.05)
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()

plt.savefig(
    "what_if_accuracy.png",
    dpi=300
)

plt.show()


# CELL 18: WHAT-IF F1 SCORE GRAPH

plt.figure(figsize=(11, 6))

sns.lineplot(
    data=what_if_df,
    x="Features",
    y="F1 Score",
    hue="Algorithm",
    marker="o"
)

plt.title(
    "Effect of TF-IDF Feature Count on F1 Score"
)

plt.xlabel("Number of TF-IDF Features")
plt.ylabel("F1 Score")

plt.ylim(0, 1.05)
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()

plt.savefig(
    "what_if_f1_score.png",
    dpi=300
)

plt.show()


# CELL 19: IDENTIFY BEST CLASSIFIER

best_model_row = results_df.loc[
    results_df["F1 Score"].idxmax()
]

best_model_name = best_model_row["Algorithm"]

print("\n================================")
print("BEST PERFORMING CLASSIFIER")
print("================================")

print("Algorithm:", best_model_name)

print(
    "Accuracy:",
    round(best_model_row["Accuracy"], 4)
)

print(
    "Precision:",
    round(best_model_row["Precision"], 4)
)

print(
    "Recall:",
    round(best_model_row["Recall"], 4)
)

print(
    "F1 Score:",
    round(best_model_row["F1 Score"], 4)
)

print(
    "\nBest model selected based on highest test F1-score."
)


# CELL 20: TEST WITH CUSTOM PRODUCT REVIEWS

print("\n================================")
print("CUSTOM PRODUCT REVIEW PREDICTION")
print("================================")

custom_reviews = [

    "This product is excellent and I really love it",

    "Very poor quality and completely disappointed",

    "The product works perfectly and delivery was fast",

    "Worst purchase ever. The product stopped working",

    "Good value for money and excellent quality"

]

# Use the original TF-IDF vectorizer
custom_features = tfidf.transform(
    [preprocess_text(review) for review in custom_reviews]
)

# Use the best-performing trained classifier
best_model = trained_models[best_model_name]

custom_predictions = best_model.predict(
    custom_features
)

for review, prediction in zip(
    custom_reviews,
    custom_predictions
):

    sentiment = (
        "Positive" if prediction == 1
        else "Negative"
    )

    print("\nReview:", review)
    print("Predicted Sentiment:", sentiment)


# CELL 21: SAVE ALL IMPORTANT OUTPUTS

print("\nSaving project outputs...")

df.to_csv(
    "processed_amazon_reviews.csv",
    index=False
)

results_df.to_csv(
    "final_comparison_results.csv",
    index=False
)

what_if_df.to_csv(
    "final_what_if_results.csv",
    index=False
)

print("\n================================")
print("PROJECT EXECUTION COMPLETED!")
print("================================")

print("\nGenerated Files:")
print("1. amazon_reviews_sample.csv")
print("2. processed_amazon_reviews.csv")
print("3. model_comparison.csv")
print("4. final_comparison_results.csv")
print("5. final_what_if_results.csv")
print("6. performance_comparison.png")
print("7. accuracy_comparison.png")
print("8. confusion_matrices.png")
print("9. what_if_accuracy.png")
print("10. what_if_f1_score.png")

print("\nAll four classifiers have been evaluated.")
```
